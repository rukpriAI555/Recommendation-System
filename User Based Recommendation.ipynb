# Import Libraries

```python
import pandas as pd
import numpy as np
```

# Load the Dataset

```python
r_cols = ['user_id', 'movie_id', 'rating', 'unix_timestamp']
ratings_df = pd.read_csv('u.data', sep="\t", names=r_cols, encoding='latin-1')
```

# Create pivot table for user and movie based on Ratings

```python
data_pivot = ratings_df.pivot_table(index='user_id', columns='movie_id', values='rating')
```

```python
data_pivot
```

# Replace NAN with zeroes

```python
data_matrix = data_pivot.replace(np.nan,0)
```

```python
data_matrix
```

# Find Cosine Similarity for User and Item

```python
from sklearn.metrics.pairwise import pairwise_distances
user_similarity = pairwise_distances(data_matrix, metric='cosine')
item_similarity = pairwise_distances(data_matrix.T, metric='cosine')
```

```python
user_similarity
```

```python
item_similarity
```

# Calculating the score value for User and Item

```python
def predict(ratings, similarity, type='user'):
    if type == 'user':
        mean_user_rating = ratings.mean(axis=1)
        #We use np.newaxis so that mean_user_rating has same format as ratings (3 by 3 matrix)
        mean_user_rating = mean_user_rating.to_numpy()
        ratings_diff = ratings - mean_user_rating[:, np.newaxis]
        pred = mean_user_rating[:, np.newaxis] + similarity.dot(ratings_diff) / np.array([np.abs(similarity).sum(axis=1)]).T
    elif type == 'item':
        pred = ratings.dot(similarity) / np.array([np.abs(similarity).sum(axis=1)])
    return pred
```

```python
#prediction table
user_prediction = predict(data_matrix, user_similarity, type='user')
item_prediction = predict(data_matrix, item_similarity, type='item')
```
