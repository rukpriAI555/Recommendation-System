# Import Libraries

```python
import pandas as pd
import numpy as np
```

# Load the Movie Lens Dataset

```python
# Define column names for ratings dataset
r_cols = ['user_id', 'movie_id', 'rating', 'unix_timestamp']

# Read ratings dataset
ratings = pd.read_csv(
    'u.data',
    sep='\t',
    names=r_cols,
    encoding='latin-1'
)

# Display ratings
ratings.head()
```

# Load Movie dataset

```python
# Read movie dataset
m_cols = [
    'movie_id', 'title', 'release_date', 'video_release_date',
    'IMDb_URL', 'unknown', 'Action', 'Adventure', 'Animation',
    'Children', 'Comedy', 'Crime', 'Documentary', 'Drama',
    'Fantasy', 'Film-Noir', 'Horror', 'Musical', 'Mystery',
    'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western'
]

movies = pd.read_csv(
    'u.item',
    sep='|',
    names=m_cols,
    encoding='latin-1'
)

movies.head()
```

# Calculate Average ratings and number of ratings - groupby movie_id

```python
movie_rating = ratings.groupby('movie_id').agg(
    mean_rating=('rating', 'mean'),
    rating_count=('rating', 'count')
).reset_index()

movie_rating.head()
```

# Merge movie details with rating

```python
movie_data = pd.merge(
    movies,
    movie_rating,
    on='movie_id'
)

movie_data.head()
```

# Find top 5 movies from every genre

```python
# List of genre columns
genre_cols = m_cols[5:]

# Create an empty dictionary
top_movies = {}

# Loop through every genre
for genre in genre_cols:

    # Select movies belonging to the genre
    genre_movies = movie_data[movie_data[genre] == 1]

    # Sort movies based on average rating
    genre_movies = genre_movies.sort_values(
        by=['mean_rating', 'rating_count'],
        ascending=False
    )

    # Select top 5 movies
    top_movies[genre] = genre_movies[
        ['title', 'mean_rating', 'rating_count']
    ].head(5)

    print("\nTop 5 movies in", genre)
    print(top_movies[genre])
```
